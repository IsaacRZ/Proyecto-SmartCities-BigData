```python
# Celda 1: Importar librerías y configuración
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import glob
import os

# Configuración de visualización para todos los plots
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("✅ Librerías importadas correctamente")
```

```python
# Celda 2: Cargar archivos CSV
# Ajusta la ruta según tu estructura de carpetas
data_path = "./data/raw/barcelona_traffic/"

# Listar todos los CSVs disponibles
csv_files = glob.glob(os.path.join(data_path, "*.csv"))
print(f"📁 Archivos encontrados: {len(csv_files)}")
for f in csv_files:
    print(f"  - {os.path.basename(f)}")
```

```python
# Celda 3: Cargar muestra de 10K registros
# Primero lee un archivo para ver la estructura
sample_file = csv_files[0]  # Toma el primero
print(f"📖 Leyendo archivo: {os.path.basename(sample_file)}")

# Leer primeras 100 filas para inspeccionar
df_peek = pd.read_csv(sample_file, nrows=100)
print(f"\n🔍 Primeras columnas encontradas:")
print(df_peek.columns.tolist())
print(f"\n📊 Primeras 5 filas:")
print(df_peek.head())
```

```python
# Celda 4: Cargar 10K registros completos
# Usa el parámetro nrows para limitar a 10K
df = pd.read_csv(sample_file, nrows=10000)

print(f"✅ Muestra cargada: {len(df):,} registros")
print(f"📏 Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
print(f"💾 Memoria usada: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
# Celda 5: Inspeccionar columnas y tipos de datos
print("📋 INFORMACIÓN DEL DATASET")
print("="*60)
df.info()

print("\n📊 TIPOS DE DATOS:")
print(df.dtypes)

print("\n🔑 NOMBRES DE COLUMNAS:")
for i, col in enumerate(df.columns, 1):
    print(f"  {i}. {col}")
```

```python
# Celda 6: Identificar campos según convenciones comunes
# Los nombres pueden variar, ajusta según lo que veas

# Posibles nombres para campos clave (ajusta según tu dataset):
timestamp_candidates = [col for col in df.columns if any(
    keyword in col.lower() 
    for keyword in ['data', 'fecha', 'time', 'timestamp', 'dia', 'hora']
)]

sensor_candidates = [col for col in df.columns if any(
    keyword in col.lower() 
    for keyword in ['id', 'sensor', 'codi', 'estacio', 'detector']
)]

vehicles_candidates = [col for col in df.columns if any(
    keyword in col.lower() 
    for keyword in ['vehicles', 'intensitat', 'volum', 'trafic', 'count']
)]

speed_candidates = [col for col in df.columns if any(
    keyword in col.lower() 
    for keyword in ['velocitat', 'speed', 'veloc']
)]

print("🔍 CAMPOS IDENTIFICADOS:")
print(f"\n⏰ Timestamp candidates: {timestamp_candidates}")
print(f"🆔 Sensor ID candidates: {sensor_candidates}")
print(f"🚗 Num Vehicles candidates: {vehicles_candidates}")
print(f"⚡ Speed candidates: {speed_candidates}")
```
